```python
import pandas as pd
from io import StringIO

lines = raw_text.strip().split('\n')
models = lines[0].split()[1:]
metrics = lines[1].split()[1:]

data = []
current_dataset = None

for line in lines[2:]:
    if not line[0].isdigit():  # It's a dataset name
        current_dataset = line.strip()
    else:
        parts = line.strip().split()
        horizon = int(parts[0])
        values = list(map(float, parts[1:]))

        for i, model in enumerate(models):
            mse = values[i * 2]
            mae = values[i * 2 + 1]
            data.append({
                'Dataset': current_dataset,
                'Horizon': horizon,
                'Model': model,
                'MSE': mse,
                'MAE': mae
            })

df = pd.DataFrame(data)
df
```
